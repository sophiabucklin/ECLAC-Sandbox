```python
s = df["periodo"].astype("Int64").astype(str).str.zfill(8) #separating the date format into YEAR MONTH WEEK

df["year"] = s.str[:4].astype(int)
df["month"] = s.str[4:6].astype(int)
df["week"] = s.str[6:8].astype(int)
```
